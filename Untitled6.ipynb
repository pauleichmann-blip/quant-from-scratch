Guessing Game

```python
import random

def main():
  number = getn()
  g_num = generate_number(number)
  while True:
    user_guess = prompt()
    if user_guess < g_num:
      print("Too small!")
    elif user_guess > g_num:
      print("Too large!")
    else:
      print("Just right!")
      break



def getn():
  while True:
    try:
      n = int(input("Level: "))
      if n > 0:
        return n
      else:
        continue
    except ValueError:
      continue

def generate_number(x):
  return random.randint(1, x)

def prompt():
  while True:
    try:
      guess = int(input("Guess: "))
      if guess > 0:
        return guess
      else:
        continue
    except ValueError:
      continue

main()
```

Little Professor

```python
import random


def main():
  user_level = get_level()
  score = 0
  for _ in range(10):
    error = 0
    x = generate_integer(user_level)
    y = generate_integer(user_level)
    while True:
      try:
        z = int(input(f"{x} + {y} = "))
        if z == x + y:
          score = score + 1
          break
        elif error == 2:
          print("EEE")
          print(f"{x} + {y} = {x + y}")
          break
        else:
          print("EEE")
          error = error + 1
          continue
      except ValueError:
        print("EEE")
        error = error + 1
        if error == 3:
            print(f"{x} + {y} = {x + y}")
            break
        continue
  print(f"Score: {score}")


def get_level():
    while True:
      try:
        level_prompt = int(input("Level: "))
        match level_prompt:
          case 1 | 2 | 3:
            return level_prompt
      except ValueError:
        continue


def generate_integer(level):
  if level == 1:
    return random.randint(0,9)
  elif level == 2:
    return random.randint(10,99)
  elif level == 3:
    return random.randint(100,999)
  else:
    raise ValueError


if __name__ == "__main__":
    main()
```

Bitcoin Price Index

```python
import requests

def main():
  try:
    n = get_value()
    from google.colab import userdata
    API = userdata.get('API_Key_CoinCap')
    request = requests.get("https://rest.coincap.io/v3/assets/bitcoin?apiKey=" + API)
    o = request.json()
    print(f"${float(o["data"]["priceUsd"]) * n:,.4f}")

  except requests.RequestException:
    pass

def get_value():
  while True:
    try:
      return float(input("Number of Bitcoin wanted: "))
    except ValueError:
      print("Wrong input")


main()
```

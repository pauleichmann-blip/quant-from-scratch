Lecture 3 Exceptions

```python
try:
  x = int(input("Whats x? "))
  print(f"x is {x}")
except ValueError:
  print("x is not an integer")
```

```python
try:
  x = int(input("Whats x? "))
except ValueError:
  print("x is not an integer")
else:
  print(f"x is {x}")
```

```python
while True:
  try:
    x = int(input("Whats x? "))
    break
  except ValueError:
    print("x is not an integer")

print(f"x is {x}")
```

```python
def main():
  x = get_int("Whats x? ")
  print(f"x is {x}")


def get_int(prompt):
  while True:
    try:
      return int(input(prompt))
    except ValueError:
      pass

main()
```

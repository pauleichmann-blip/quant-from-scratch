Liste gegen Array

Vorhersage:
print(liste * 2) -> 1 2 3 1 2 3
print(a * 2) -> 2 4 6
print(liste + [10, 20, 30]) -> 1 2 3 10 20 30
print(a + np.array([10, 20, 30])) -> 11 22 33
print(a ** 2) -> 100 400 900

```python
import numpy as np

liste = [1, 2, 3]
a = np.array([1, 2, 3])

print(liste * 2)
print(a * 2)
print(liste + [10, 20, 30])
print(a + np.array([10, 20, 30]))
print(a ** 2)
```

Vorhersagen:
print(a[0]) -> 4
print(a[-1]) -> 42
print(a[1:3]) -> 8 15 16
print(a > 10) -> False
print(a[a > 10]) -> 15 16 23 42
print((a > 10).sum()) -> 96

```python
a = np.array([4, 8, 15, 16, 23, 42])

print(a[0])
print(a[-1])
print(a[1:3])
print(a > 10)
print(a[a > 10])
print((a > 10).sum())
```

Vorhersage:
print(km.sum()) -> 10
print(km.mean()) -> 2
print((km > 0).mean()) -> 3
print(np.cumsum(km)) -> keine Ahnung
print(km.std()) -> 2.12

```python
km = np.array([3, 0, 5, 2, 0])

print(km.sum())
print(km.mean())
print((km > 0).mean())
print(np.cumsum(km))
print(km.std())
```

```python
km = np.array([3, 0, 5, 2, 0])
print(((km - km.mean())**2).mean() ** 0.5)
print(km.std(ddof=1))
```

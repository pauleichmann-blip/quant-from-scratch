camelCase

```python
def main():
  camel = input("camelCase: ")
  split(camel)

def split(n):
  print("snake_case: ", end="")
  for i in n:
    if i.islower():
      print(i, end="")
    elif i.isupper():
      print(f"_{i.lower()}", end="")
    else:
      print(f"_{i}", end="")
  print("")

main()
```

Coke Machine

```python
def main():
  getcoins()

def getcoins():
  price = 50
  print(f"Amount Due: {price}")
  while price > 0:
    x = int(input("Insert Coin: "))
    match x:
      case 5 | 10 | 25:
        price = price - x
        check_price(price)
      case _:
        print(f"Amount Due: {price}")

def check_price(n):
  if n <= 0:
    print(f"Change Owed: {-n}")
  else:
    print(f"Amount Due: {n}")

main()
```

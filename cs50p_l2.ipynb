camelCase

```python
def main():
  camel = input("camelCase: ")
  split(camel)

def split(n):
  print("snake_case: ", end="")
  for i in n:
    if i.islower():
      print(i, end="")
    elif i.isupper():
      print(f"_{i.lower()}", end="")
    else:
      print(f"_{i}", end="")
  print("")

main()
```

Coke Machine

```python
def main():
  getcoins()

def getcoins():
  price = 50
  print(f"Amount Due: {price}")
  while price > 0:
    x = int(input("Insert Coin: "))
    match x:
      case 5 | 10 | 25:
        price = price - x
        check_price(price)
      case _:
        print(f"Amount Due: {price}")

def check_price(n):
  if n <= 0:
    print(f"Change Owed: {-n}")
  else:
    print(f"Amount Due: {n}")

main()
```

2. Versuch Vanity Plates

```python
def main():
  plate = input("Plate: ")
  if is_valid(plate):
    print("Valid")
  else:
    print("Invalid")


def is_valid(s):
  if max_6_char(s) and two_letters(s) and no_spec_char(s) and order(s):
    return True
  else:
    return False



def two_letters(l):
    if l[0].isalpha() == False or l[1].isalpha() == False:
      return False
    else:
      return True


def max_6_char(c):
  if 2 <= len(c) <= 6:
    return True
  else:
    return False


def no_spec_char(no):
  if no.isalnum():
    return True
  else:
    return False


def order(p):
  if p.isalpha():
    return True
  else:
    for x in range(len(p)):
      if p[x].isdigit():
        if p[x] == "0":
          return False
        elif p[x:].isdigit():
          return True
        else:
          return False



main()
```
